```python
import pandas as pd

df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'], 
                    'B': ['B0', 'B1', 'B2', 'B3'], 
                    'C': ['C0', 'C1', 'C2', 'C3'], 
                    'D': ['D0', 'D1', 'D2', 'D3']}, 
                  index=[0,1,2,3])
```

```python
df1
```

```python
df2 = pd.DataFrame({'A': ['A4', 'A5', 'A6', 'A7'], 
                    'B': ['B4', 'B5', 'B6', 'B7'], 
                    'C': ['C4', 'C5', 'C6', 'C7'], 
                    'D': ['D4', 'D5', 'D6', 'D7']}, 
                  index=[4,5,6,7])
```

```python
df2
```

 loc vs iloc
 

```python
s = pd.Series(list("abcdef"), index=[49, 48, 47, 0, 1, 2]) 
s
```

```python
s.loc[0] # i.e locate the value of label of key 0
```

```python
s.loc[49] # locate the value of label of key 49
```

```python
s.loc[0:1]    # locate the value of label of key 0 to 1
```

```python
s.loc[49:48]   # locate the value of label of key 49 to 48
```

```python
s.iloc[0] # index location not label 
```

```python
s.iloc[0:5]  # index location not label 
```

sort the series 

```python
s = s.sort_index() # this will rearrange the series 
```

```python
s
```

```python
# now let apply index location 
s.iloc[0]
```

```python
s.iloc[0:2]
```

```python
# let see label
s.loc[0]
```

```python
s.loc[0:2] # this is because it has been arranged 
```

```python
s.loc[0:49] # location dealas with label and it will start counting from 1 not 0
```

```python
s.iloc[0:6] # this start from index of 0 to 5 hence the 6
```
